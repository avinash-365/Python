# Merging Joining and Concatination

## Merging 2 dataframes

```python
data1 = {
    "Emp_ID": [101, 102, 103, 104],
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 22, 28]
}
df1 = pd.DataFrame(data1)


data2 = {
    "Emp_ID": [101, 102, 103, 105],
    "Department": ["HR", "Engineering", "Marketing", "Sales"],
    "City": ["New York", "San Francisco", "London", "Chicago"]
}
df2 = pd.DataFrame(data2)

df1
```

```python
df2
```

```python
pd.merge(df1,df2) # by default hoe perfrom to inner opeartion
```

```python
pd.merge(df1,df2,on="Emp_ID",how='outer')
```

```python
pd.merge(df1,df2,on="Emp_ID",how='left')
```

```python
pd.merge(df1,df2,on="Emp_ID",how='right')
```

## Concatination of 2 dataframes

```python
df1 = pd.DataFrame({
    'Store_ID': [101, 102, 103],
    'Product': ['Apple', 'Banana', 'Orange'],
    'Sales': [250, 180, 310]
})

df1
```

```python
df2 = pd.DataFrame({
    'Store_ID': [104, 105, 106],
    'Product': ['Apple', 'Mango', 'Banana'],
    'Sales': [210, 150, 190]
})

df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df2,df1]) # base on column
```

```python
pd.concat([df1,df2],axis=1) # base on rows
```

## joining of 2 dataframes

```python
emp1 = {
    'Employee_ID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Dept_ID': [1, 2, 1, 4]  
}
df_1 = pd.DataFrame(emp1)


dep = {
    'Dept_ID': [1, 2, 3], 
    'Dept_Name': ['HR', 'Engineering', 'Marketing']
}
df_2 = pd.DataFrame(dep)

df_2
```

```python
df_1
```

```python
df_1.join(df_2,lsuffix='_emp', rsuffix='_dept')
```

### Differnce between merge and join 
• pd.merge() (SQL-style):

	• Best used when you want to match specific columns from two different tables (e.g., matching Employee_ID to Employee_ID).
	• Allows you to specify left and right columns or indices explicitly using left_on and right_on.

• df.join() (Index-style):

	• Best used when both of your datasets already share a clean, identical index (like a Date or an ID set as the index).
	• Saves keystrokes when combining a list of DataFrames that align by their row labels.


# Group By Aggregation

```python
# Group by dos'n create dataframe create object
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
df
```

```python
# Group by Category and calculate the sum of Sales
cat = df.groupby("Category")["Sales"].sum()
cat
```

```python
# Group by Store and calculate the sum of Sales
cat = df.groupby("Store")["Sales"].sum()
cat
```

```python
# Group by multiple columns
# Group by Category and Store
cat = df.groupby(["Store","Category"])["Sales"].sum()
cat
```

## Aggregation

```python
df["Sales"].mean()
df["Sales"].min()
df["Sales"].sum()
```

```python
df["Sales"].agg(['mean','sum','min','max'])
```

# PivotTable and Cross Tabs

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
```

```python
# pivottable use case hit map creation 
```

```python
pd.pivot_table(df,values="Sales",index="Region",columns="Product")
```

```python
pd.pivot_table(df,values=["Sales","Units"],index="Region",columns="Product",aggfunc='sum')
```

## Cross tab

```python
pd.crosstab(df["Region"],df["Product"])
```
