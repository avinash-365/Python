```python
import numpy as np

print("--- Algorithmic Trading Analytics Engine ---\n")

# Step 1: Data Initialization (365 divas, 5 companies)
# 50 thi 500 ni range ma random prices USD ma
array_USD = np.random.uniform(50.0, 500.0, (365, 5))

# Step 2: Data Transformation (Broadcasting)
# USD ne INR ma convert karva (1 USD = 95.5 INR)
array_2D = array_USD * 95.5
print("1. Data Shape (INR):", array_2D.shape) # Output: (365, 5)

# Step 3: Data Extraction (Slicing & Indexing)
# Company-3 (Index 2) na pehla 30 divas
company_3_month_1 = array_2D[0:30, 2]

# Badhi companies na chella 10 divas
last_10_days = array_2D[-10:, :]

# Step 4: Statistical Analysis (Aggregations using axis=0)
avg_price = np.mean(array_2D, axis=0)
max_price = np.max(array_2D, axis=0)
min_price = np.min(array_2D, axis=0)
volatility = np.std(array_2D, axis=0)

print("\n--- Statistics (Per Company) ---")
print("Average Prices:", np.round(avg_price, 2))
print("Volatility (Risk):", np.round(volatility, 2))

# Step 5: Advanced Filtering (Boolean Masking)
# Company-1 (Index 0) no price 25000 (INR) karta vadhare hoy teva divaso
mask = array_2D[:, 0] > 25000
high_price_days_count = np.sum(mask)
print(f"\nCompany-1 stock crossed 25000 INR on {high_price_days_count} days.")

# Step 6: Finding Patterns (argmax)
# Company-2 (Index 1) no highest price kaya divase hto?
highest_day_index = np.argmax(array_2D[:, 1])
print(f"Company-2 hit its maximum price on day index: {highest_day_index}")

# Step 7: Reshaping Data
# 365 ne 7 thi bhagi nathi shakata, etle pehla 364 divas alag karya
data_364 = array_2D[:364, :]
# Have tene 52 weeks x 7 days x 5 companies ma reshape karyu
weekly_data = np.reshape(data_364, (52, 7, 5))
print("\nReshaped Weekly Data Shape:", weekly_data.shape) # Output: (52, 7, 5)

# Step 8: Real-world Metrics (Daily Return Percentage without loops)
# Offset Slicing logic
todays_prices = array_2D[1:, :]      # Divas 2 thi chhek end sudhi
yesterdays_prices = array_2D[:-1, :]   # Divas 1 thi second-last sudhi

# Formula: ((Today - Yesterday) / Yesterday) * 100
daily_return = ((todays_prices - yesterdays_prices) / yesterdays_prices) * 100
print("\nDaily Returns Shape:", daily_return.shape) # Output: (364, 5)
print("Daily Returns (First 2 days, all companies):\n", np.round(daily_return[:2, :], 2))
```
