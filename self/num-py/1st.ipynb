## Numpy Basic and Array

```python
import numpy as np
```

## Creating array from list

```python
arr_1d = np.array([1,2,3])
print("1D array:",arr_1d)

arr_2d = np.array([[1,2,3],[4,5,6]])
print("2D array:",arr_2d)
```

## List Vs Array

```python
py_list = [1,2,3]
print("Multiplication of list",py_list * 2)

np_array = np.array([1,2,3]) # element wise multiplication
print("Multiplication of array",np_array * 2)

import time

start = time.time()
py_list = [i*2 for i in range(1,100000)]
print("\nList operation time:",time.time() - start)

start = time.time()
np_array = np.arange(100000) * 2
print("Array operation time:",time.time() - start)
```

### creating array from scratch

```python
zero = np.zeros((3,4))
print("Zeros array:\n",zero)

ones = np.ones((3,4))
print("Ones array:\n",ones)

full = np.full((2,2),5)
print("Full array:\n",full)

random = np.random.random((2,3))
print("Random array:\n",random)

sequnce = np.arange(0,10,2) # non-incluctive
print("Sequnce array:\n",sequnce)

Empty = np.empty(2)
print("Empty array:\n",Empty)

linspace = np.linspace(0,10,num=5)
print("array with values that are spaced linearly:\n",linspace)
```

### Vector , Matrix and Tensor

```python
vector = np.array([1,2,3])
print("Vector:",vector)

matrix = np.array([[1,2,3] , [4,5,6]])
print("Matrix:",matrix)

tensor = np.array([[[1,2],[3,4]],
                    [[5,6],[7,8]]])
print("Tensor:",tensor)
```

### Array Properties

```python
arr = np.array([[1,2,3],
               [4,5,6]])

print("Size" , arr.size)
print("Shape" ,arr.shape)
print("Dimension ",arr.ndim)
print("dtype ",arr.dtype)
```

```python
arr = np.arange(12)
print("Original array ",arr)

reshaped = arr.reshape((3, 4))
print("\nReshaped array ",reshaped)

flattened = reshaped.flatten()
print("\nFlattened array ",flattened)

# ravel (returns view , insteead of copy)
raveled = reshaped.ravel()
print("\nraveled array ",raveled)

# Transpose
transpose = reshaped.T
print("\ntranspose array ",transpose)
```

## Numpy Array operation and indexing

```python
arr = np.arange(10)

print("Original Array:",arr)
print("Basic Slicing:",arr[3:8])
print("With Step:",arr[1:8:2])
print("Negative indexing:",arr[-3:-1])
```

```python
arr_2d = np.random.randint(10,size=(3,3))

print("Original Array:",arr_2d)
print("Specific element:",arr_2d[1,2])
print("Entire Row:",arr_2d[1])
print("Entire Column:",arr_2d[:,1])
```

### Sorting of Array

```python
unsorted = np.array([1,8,9,6,4,3,2,5,7])

print("Sorted Array:",np.sort(unsorted))

unsorted_2d = np.array([[1,3],[4,2],[6,5]])
print("Sorted Array by Column:",np.sort(unsorted_2d,axis=1)) # axis 1 means left to right
print("Sorted Array by Row:",np.sort(unsorted_2d,axis=0)) # axis 0 means top to bottom
```

## Filter and mask

```python
number = np.arange(10)

even = number[number % 2 == 0]
print("Even Numbers:",even)
```

```python
mask = number > 5
print("Numbers greater than 5 ",number[mask])
```

## Fancy indexing vs np.where()

```python
indices = [0,2,6]
print(number[indices])

where_result = np.where(number > 5)
print("NP Where ",number[where_result])

condition_array = np.where(number > 5 ,number * 2 , number)   # where (Conditon,True part , False part)
print(condition_array)
```

## Adding and Removing data

```python
arr1 = [1,2,3]
arr2 = [4,5,6]

combine = np.concatenate((arr1,arr2))
print("Combine Array:",combine)
```

```python
a = np.array([1,2,3])
b = np.array([1,2,3])

print("Compatibility shape :", a.shape == b.shape)


original = np.array([[1,2],[3,4]])
new_row = np.array([[4,5]])

with_new_row = np.vstack((original,new_row)) # v-stack always add row , row left to right 
print(original)
print()
print(with_new_row)


new_col = np.array([[7],[8]])
with_new_col = np.hstack((original,new_col)) # h-stack new column add
print("With new column ",with_new_col)

arr= np.array([1,2,3,4,5])
deleted = np.delete(arr,2)
print()
print("Array after deleted :",deleted)
```
